```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('car_dataset.csv')
df.columns
```

5. Show top 5 most frequent car models.

```python
frequent_car = df.groupby('Model')['Year'].count()
frequent_car.sort_values(ascending=False).head(5)
```

6. Mean, median, min, max of Price.

```python
print('mean of price:-',df['Price'].mean())
print('median of price:-',df['Price'].median())
print('min of price:-',df['Price'].min())
print('max of price:-',df['Price'].max())
```

7. Min/max mileage and engine CC.

```python
df.describe()
```

10. Histogram of Car Prices.

```python
plt.figure(figsize=(10,5))
sns.histplot(x='Model',y='Price',data=df,hue='Model')
plt.xticks(rotation=90)
plt.show()
```

11. Histogram of Mileage.

```python
plt.figure(figsize=(10,5))
sns.histplot(x='Model',y='Mileage',data=df,hue='Model',palette='tab10')
plt.xticks(rotation=90)
plt.show()
```

12. Bar chart – Cars count per Brand.

```python
plt.figure(figsize=(10,5))
cars_per_brand=df.groupby('Brand')['Model'].count().reset_index()
ax=sns.barplot(x='Brand',y='Model',data=cars_per_brand,hue='Brand')
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

13. Pie chart – Fuel Type Distribution.

```python
FuelType=df.groupby('Fuel_Type')['Model'].count().reset_index()
plt.figure(figsize=(6,6))
plt.pie(FuelType['Model'],labels=FuelType['Fuel_Type'],autopct='%1.1f%%')
plt.show()
```

14. Countplot – Transmission type.

```python
plt.figure(figsize=(10,5))
ax = sns.countplot(x='Transmission',data=df,hue='Transmission')
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```


15. Boxplot – Service Cost.

```python
plt.figure(figsize=(20,10))
ax = sns.barplot(data=df,x='Model',y='Service_Cost',hue='Model')
plt.xticks(rotation=90)
plt.show()
```

16. Bar chart – Number of cars per Year.

```python
Num_of_cars=df.groupby('Year')['Model'].count().reset_index()
plt.figure(figsize=(10,5))
ax = sns.barplot(x='Year',y='Model',hue='Year',data=Num_of_cars)
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

17. Countplot – Seating Capacity.

```python
plt.figure(figsize=(10,5))
ax = sns.countplot(x='Seating_Capacity',data=df,hue='Seating_Capacity')
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

18. Bar chart – Top 10 car models.

```python
topCars = df.groupby('Model')['Brand'].count().sort_values(ascending=False).head(10).reset_index()
plt.figure(figsize=(10,5))
ax=sns.barplot(x='Model',y='Brand',hue='Model',data=topCars,palette='Dark2')
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

```python
plt.figure(figsize=(10,5))
ax = sns.boxplot(x='Transmission',y='Price',data=df)
plt.show()
```

22. Bar chart – Avg Service Cost by Brand.

```python
avgService = df.groupby('Brand')['Service_Cost'].mean().reset_index()
plt.figure(figsize=(10,5))
ax = sns.barplot(x='Brand',y='Service_Cost',data=avgService,hue='Brand')
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

23. Bar chart – Avg Mileage by Fuel Type.

```python
Avgmil = df.groupby('Fuel_Type')['Mileage'].mean().reset_index()
plt.figure(figsize=(10,5))
ax = sns.barplot(x='Fuel_Type',y='Mileage',data=Avgmil,hue='Fuel_Type')
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```


24. Line chart – Avg Price per Year.

```python
from turtle import color


AvgPrice = df.groupby('Year')['Price'].mean().reset_index()
plt.figure(figsize=(10,5))
ax = sns.lineplot(x='Year', y='Price', data=AvgPrice,color='Red')
plt.title('Average Price Trend by Year')
plt.show()
```

25. Bar chart – Avg Engine CC by Brand.

```python
from sympy import rotations


AvgEngine = df.groupby('Brand')['Engine_CC'].mean().reset_index()
plt.figure(figsize=(10,5))
ax = sns.barplot(x='Brand',y='Engine_CC',hue='Brand',data=AvgEngine)
for container in ax.containers:
    ax.bar_label(container)
plt.title('Average Engine_CC for each brand')
plt.xticks(rotation=45)
plt.show()
```

26. Correlation heatmap.

```python
# Select only numeric columns
correlation_Matrix = df.select_dtypes(include=['number']).corr()

# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_Matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix Heatmap", fontsize=16)
plt.show()
```

33. Heatmap – Brand vs Fuel Type.

```python
brand_vs_fuel = pd.crosstab(df['Brand'], df['Fuel_Type'])
sns.heatmap(brand_vs_fuel, cmap='coolwarm', annot=True, fmt='d')
plt.show()
```

34. Swarmplot – Engine_CC vs Brand.

```python
sns.swarmplot(y='Engine_CC',x='Brand',data=df,size=7,hue='Brand')
plt.show()
```

35. Stacked Bar – Transmission per Brand.

```python
from matplotlib.pyplot import bar_label


Tr_per_brand = df.groupby(['Brand','Transmission'])['Car_ID'].count().reset_index()
pivotTr = Tr_per_brand.pivot(index='Brand',columns='Transmission',values='Car_ID')
pivotTr.plot(kind='bar',stacked=True,figsize=(10,5))
```
